```python
from pathlib import Path
import pandas as pd

# Define the dataset location
file_path = Path("../data/raw/DataCoSupplyChainDataset.csv")

# Load the CSV
df = pd.read_csv(
    file_path,
    encoding="latin1",
    low_memory=False
)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# List all columns
print(df.columns.tolist())

# Check missing values
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])

# Check unique orders
print("Unique orders:", df["Order Id"].nunique())

# Check delivery statuses
print(df["Delivery Status"].value_counts())
```

Let's calculate a preliminary delivery performance metric at the record level.

```python
# Exclude canceled records
completed = df[
    df["Delivery Status"] != "Shipping canceled"
].copy()

# Count on-time and early records
on_time = completed[
    completed["Delivery Status"].isin([
        "Shipping on time",
        "Advance shipping"
    ])
]

# Calculate on-time delivery percentage
otd_rate = len(on_time) / len(completed) * 100

print(f"Preliminary on-time delivery rate: {otd_rate:.2f}%")
```
